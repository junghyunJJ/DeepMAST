```python
from src.agents.deep_interpreter.tools.data_loader import (
    load_analysis_data,
    get_recommended_targets,
    ANALYSIS_FILES,
)
```

## 1. 함수 실행

```python
# 데이터 디렉토리 지정 (res/ 폴더 경로)
DATA_DIR = "../res"  # notebook/ 폴더에서 상위의 res/ 참조

# 함수 실행
result = load_analysis_data.invoke({"data_dir": DATA_DIR})

print(f"반환된 키: {list(result.keys())}")
```

## 2. 기본 정보

```python
print("=" * 60)
print("기본 정보")
print("=" * 60)
print(f"data_dir: {result['data_dir']}")
print(f"available_analyses: {len(result['available_analyses'])}개")
print(f"missing_analyses: {len(result['missing_analyses'])}개")
```

## 3. 분석 파일 매핑

```python
print("=== Available Analyses ===")
for name in result['available_analyses']:
    print(f"  ✓ {name}")

if result['missing_analyses']:
    print("\n=== Missing Analyses ===")
    for name in result['missing_analyses']:
        print(f"  ✗ {name}")
```

```python
print("=== Figures ===")
for k, v in result['figures'].items():
    print(f"  {k}: {v}")
```

```python
print("=== Data (CSV) ===")
for k, v in result['data'].items():
    print(f"  {k}: {v}")
```
